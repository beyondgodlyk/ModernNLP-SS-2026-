#### Functions

```python
# Define a function that tokenizes text by splitting on whitespace (with a docstring)
def naive_tokenizer(sentence):
    return sentence.split()

# Test it on a sample sentence
sample_sentence = "There are not many, people in the Practice Session !    It is sunny!"
tokens = naive_tokenizer(sample_sentence)
print(f"Tokens are: {tokens}")
```

#### NumPy for NLP - Bag of words example

```python
# Create a simple corpus
corpus = [
    "the cat sat on the mat",
    "the dog sat on the log",
    "cats and dogs are pets"
]

# Build the vocabulary (set of all words) and a word-to-index dictionary
vocab = set()
for sentence in corpus:
    for token in naive_tokenizer(sentence):
        vocab.add(token)
print(f"Vocab is: {vocab}")
token_to_idx = {token:i for i, token in enumerate(vocab)}
print(f"Vocab size: {len(vocab)}")
print(f"Mapping: \n{token_to_idx}")

# Define a function that creates a bag-of-words vector (np.zeros + word counts) for a sentence
def create_bow(sentence, vocab, token_to_idx):
    bow = np.zeros(len(vocab))
    for token in naive_tokenizer(sentence):
        bow[token_to_idx[token]] += 1
    return bow

# Create the BoW vector for the first sentence and print it with its shape
sentence1 = corpus[0]
bow_vector1 = create_bow(sentence1, vocab, token_to_idx)

print(f"Sentence: {sentence1}")
print(f"BoW vector: {bow_vector1}")
print(f"Shape of BoW vector: {bow_vector1.shape}")

# Create the document-term matrix for the whole corpus and print its shape
```

#### Differentiation Operations with PyTorch

```python
# Create a scalar tensor x = 2.0 with requires_grad=True
x = torch.tensor(2.0, requires_grad=True)
print(f"x = {x}, requires_grad={x.requires_grad}")

# Define y = x^2 + 3x + 1, call backward() and print x.grad (expected: 7)
y = x**2 + 3*x + 1
y.backward()
print(f"dy/dx at x=2: {x.grad}")

# Vector example: x = [1, 2, 3], y = (x ** 2).sum(), backward, print gradients (expected: [2, 4, 6])
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
y = (x**2).sum()
y.backward()
print(x.grad)

# Detach a tensor to stop gradient tracking and compare requires_grad
x = torch.tensor(2.0, requires_grad=True)
y = x ** 2
z = x.detach()
print(f"Does z requires grad: {z.requires_grad}")

# Compute inside a torch.no_grad() context and check requires_grad
with torch.no_grad():
    w = x * 3
    print(w.requires_grad)
```

#### Simple linear regression example and PyTorch Training Loop

```python
# Generate synthetic data
torch.manual_seed(42)
np.random.seed(42)

X = torch.linspace(0, 10, 100).reshape(-1, 1)
true_w = 2.0
true_b = 1.0
y = true_w * X + true_b + 0.5 * torch.randn(X.shape)

# Define a simple linear model (nn.Module subclass with an nn.Linear(1, 1) layer and a forward method)
class LinearModel(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = torch.nn.Linear(1, 1, bias = True)

    def forward(self, x):
        return self.linear(x)

model = LinearModel()

# Define the loss (MSELoss) and the optimizer (SGD, lr=0.01)
loss_fn = torch.nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr = 0.01)

# Training loop for 100 epochs: forward, loss, zero_grad, backward, step; print the loss every 20 epochs
for epoch in range(500):
    y_pred = model(X)
    loss = loss_fn(y_pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 20 == 0:
        print(f"Epoch {epoch+1}: Loss = {loss.item():.5f}")
```

```python
# Print the true vs learned parameters (weight and bias)
learned_w = model.linear.weight.item()
learned_b = model.linear.bias.item()
print(f"True w = {true_w}, learned w = {learned_w:.4f}")
print(f"True b = {true_b}, learned b = {learned_b:.4f}")
```
